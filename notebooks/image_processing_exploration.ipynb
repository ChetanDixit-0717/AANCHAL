```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Create a black image: 200 rows, 300 columns, 3 channels
image = np.zeros((200, 300, 3), dtype=np.uint8)

print("Shape:", image.shape)
print("Data type:", image.dtype)
print("Pixel at (0, 0):", image[0, 0])

plt.imshow(image)
plt.title("Black Image")
plt.axis("off")
```

```python
# OpenCV arrays use BGR channel order
image[50,50] = [0,0,255] # Red pixel
image[50,100] = [0,255,0] # Green pixel
image[50,150] = [255,0,0] # Blue pixel

image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.imshow(image_rgb)
plt.title("Changing Individual Pixels")
plt.axis("off")
```

```python
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

print("Original shape:", image.shape)
print("Grayscale shape:", gray_image.shape)

plt.imshow(gray_image, cmap="gray", vmin=0, vmax=255)
plt.title("Grayscale Image")
plt.axis("off")
```
